```python
import numpy as np 
```

```python
arr=np.array([10,25,29,25,20,98])
print(arr)
print(arr.shape)
print(arr.ndim)
print(len(arr))
```

```python
arr=np.array([10,25,29,20,90])
print(arr)
print(sum(arr)/len(arr))
print(arr.mean())

```

```python
################## 2D Array ##################
arr2=np.array([[23,53,53,73],[35,53,13,76]])
print(arr2)
print(arr2.shape)
print(arr2.ndim)
print(arr2.size)
```

```python
##################### 3D Array ####################
arr3=np.array([
    [
        [23,53,98],
        [89,24,25],
        [82,52,63]
    ],
    
    [
        [89,12,64],
        [90,53,29],
        [89,53,69]
     ]
])
print(arr3)
print(arr3.shape)
print(arr3.ndim)
print(arr3.size)
```

```python
############## Mathematical Operations ###################
ar1=np.array([10,20,50])
ar2=np.array([2,5,7])
np.add(ar1,ar2)
np.subtract(ar1,ar2)
np.multiply(ar1,ar2)
np.divide(ar1,ar2)
```
