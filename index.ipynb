```python
import numpy as np 
```

```python
arr=np.array([10,25,29,25,20,98])
print(arr)
print(arr.shape)
print(arr.ndim)
print(len(arr))
```

```python
arr=np.array([10,25,29,20,90])
print(arr)
print(sum(arr)/len(arr))
print(arr.mean())

```

```python
################## 2D Array ##################
arr2=np.array([[23,53,53,73],[35,53,13,76]])
print(arr2)
print(arr2.shape)
print(arr2.ndim)
print(arr2.size)
```

```python
##################### 3D Array ####################
arr3=np.array([
    [
        [23,53,98],
        [89,24,25],
        [82,52,63]
    ],
    
    [
        [89,12,64],
        [90,53,29],
        [89,53,69]
     ]
])
print(arr3)
print(arr3.shape)
print(arr3.ndim)
print(arr3.size)
```

```python
############## Mathematical Operations ###################
ar1=np.array([10,20,50])
ar2=np.array([2,5,7])
np.add(ar1,ar2)
np.subtract(ar1,ar2)
np.multiply(ar1,ar2)
np.divide(ar1,ar2)
```

```python
arr=np.array([[10,25,29,20,90],[13,52,51,55,78]])
newLt=[]
newlt2=np.array([newLt])
for rows in arr:
        for l in rows:
            newLt.append(l)

print(newlt2)
print(type(arr))
print(type(newlt2))
np.mean(newLt)
np.median(newlt2)
```

```python
############## Stastical Operations ################

a1=np.array([23,52,89,25,8,7])
np.max(a1)
np.min(a1)
np.sum(a1)
np.mean(a1)
np.median(a1)

```

```python
##### reshape: convert 1d array into 2d ##########
ar1=np.array([23,5,7,8,9,3])
nar1=ar1.reshape(2,3)
print(nar1)
```

```python
############ flatten: convert 2d array into 1d array ##################
ar2=np.array([[23,5,9,2,6],[25,8,9,78,9]])
print(ar2.shape)
newAr=ar2.flatten()
print(newAr)
```

```python
########## converting 2d array into 1d using loop ###########
ar2=np.array([[23,5,9,2,6],[25,8,9,78,9]])
newlist=[]
for rows in ar2:
    for value in rows:
        newlist.append(value)
print(newlist)
```

```python
ar1=np.array([[1,2,3],[7,8,3],[10,11,12]])
ar2=np.array([[4,3,5],[8,7,6],[12,11,10]])

product=np.dot(ar1,ar2)
print(product)
```
