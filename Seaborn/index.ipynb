```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
titanic=sns.load_dataset("titanic")
```

```python
grades={
    "subject":["English","English","Math","Math","Urdu","Urdu"],
    "Gender":["Male","Female","Male","Female","Male","Female"],
    "Marks":[64,78,70,34,61,88]
    }
grades
```

```python
data=pd.DataFrame(grades)
```

```python
data
```

```python
sns.barplot(x="subject",y="Marks",hue="Gender",data=data)
```

```python
sns.stripplot(x="sex",y="age",hue="pclass",data=titanic,palette="rainbow")
```

```python
sns.swarmplot(x="sex",y="age",hue="pclass",data=titanic,palette="rainbow")
```
